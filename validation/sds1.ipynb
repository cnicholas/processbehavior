# SDS1 Validation Notebook

Testing human-readable residual aliases with SDS1 validation data.

```python
from processbehavior import ProcessBehavior

pb = ProcessBehavior.read_csv('sds1_data.csv')
print('Columns:', pb.cols)
pb.data.head()
```

## Formulation 1: Three factors including time

response=y, time=time, factors=[factor 1, factor 2, time]

```python
study1 = pb.formulate(
    response='y',
    time='time',
    factors=['factor 1', 'factor 2', 'time']
)
study1
```

```python
# R5_Xbar
result = study1.execute(chart='R5_Xbar')
result.plot(chart='R5_Xbar')
```

```python
# R5_S
result = study1.execute(chart='R5_S')
result.plot(chart='R5_S')
```

## Formulation 2: Two factors + time

response=y, time=time, factors=[factor 1, factor 2]

```python
study2 = pb.formulate(
    response='y',
    time='time',
    factors=['factor 1', 'factor 2']
)
study2
```

```python
# R5_Xbar using numeric notation
result = study2.execute(chart='R5_Xbar')
result.plot(chart='R5_Xbar')
```

```python
# R5_S using numeric notation
result = study2.execute(chart='R5_S')
result.plot(chart='R5_S')
```

```python
# noise_Xbar using human-readable alias (maps to R5)
result = study2.execute(chart='noise_Xbar')
print('Charts in result:', list(result.charts.keys()))
result.plot(chart='R5_Xbar')
```

```python
# noise_S using human-readable alias
result = study2.execute(chart='noise_S')
result.plot(chart='R5_S')
```
