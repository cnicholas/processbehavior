# Effects and Interactions Visualization

This notebook demonstrates the new effects visualization capabilities:
- `result.plot(chart='Effects')` - Main effects bar chart
- `result.plot(chart='TimeInteraction')` - Factor × time interaction
- `result.plot(chart='FactorInteraction')` - Factor × factor interaction

Uses the PM SDS studies from the validation dataset.

```python
from processbehavior import ProcessBehavior

# Load validation data
pb = ProcessBehavior.read_csv('PBTESTDATABASE_T100.csv')
print(f"Columns: {pb.cols}")
```

## Study 1: PM SDS 1 (Full Replication)

SDS 1 has full replication, meaning multiple observations per cell.
This allows calculation of all effects and interactions.

```python
study_pm1 = pb.formulate(
    response='PM SDS 1',
    time='PRODUCTION TIME',
    factors=['FACTOR 1', 'FACTOR 2']
)
print(f'SDS: {study_pm1.sds}')

result_pm1 = study_pm1.execute()
print(f'Has effects: {result_pm1.has_effects}')
print(f'Has interactions: {result_pm1.has_interactions}')
```

### Main Effects Chart

```python
# Main effects visualization
result_pm1.plot(chart='Effects').show()
```

### Time Interaction Chart

Shows how factor effects change over time. Non-parallel lines indicate interaction.

```python
# Factor × time interaction
result_pm1.plot(chart='TimeInteraction').show()
```

### Factor Interaction Chart

Shows the two-factor interaction (FACTOR 1 × FACTOR 2).

```python
# Factor × factor interaction (heatmap)
result_pm1.plot(chart='FactorInteraction').show()
```

### Accessing Raw Effect Data

```python
# Access effects data directly
print("Effects keys:", list(result_pm1.effects.keys()))
print("\nInteractions keys:", list(result_pm1.interactions.keys()))
```

```python
# Factor 1 main effects
result_pm1.effects['FACTOR 1']
```

```python
# Time main effects
result_pm1.effects['time']
```

```python
# Factor × factor interaction
result_pm1.interactions['factor_factor']
```

## Study 2: PM SDS 2 (No Replication)

SDS 2 has no replication (one observation per cell). Uses IMR chart.

```python
study_pm2 = pb.formulate(
    response=pb.cols.PM_SDS_2,
    time='PRODUCTION TIME',
    factors=['FACTOR 1', 'FACTOR 2']
)
print(f'SDS: {study_pm2.sds}')

result_pm2 = study_pm2.execute(chart='Imr', by=[])
print(f'Has effects: {result_pm2.has_effects}')
print(f'Has interactions: {result_pm2.has_interactions}')
```

```python
# Main effects for SDS 2
result_pm2.plot(chart='Effects').show()
```

```python
# Time interaction for SDS 2
result_pm2.plot(chart='TimeInteraction').show()
```

```python
# Factor interaction for SDS 2
result_pm2.plot(chart='FactorInteraction').show()
```

## Summary

The effects visualization API provides:

| Chart Type | Visualization | Data Source |
|------------|---------------|-------------|
| `'Effects'` | Horizontal bar chart | `effects[factor_name]`, `effects['time']` |
| `'TimeInteraction'` | Line plot over time | `interactions['factor_time']` |
| `'FactorInteraction'` | Heatmap | `interactions['factor_factor']` |

All SDS types (1, 2, 3) with factors support these visualizations.
